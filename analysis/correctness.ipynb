```python
from shared import load_outputs, load_truths, score
import pandas as pd

REPOS = ['gitignore', 'ponytail', 'markitdown', 'stable-diffusion-webui', 'scrcpy', 'iptv', 'vue', 'MoneyPrinterTurbo', 'ohmyzsh', 'clash-verge-rev']

outputs, truths = load_outputs(), load_truths()
rows = []

for (model, repo), runs in outputs.items():
    if repo not in REPOS:
        continue

    truth, raw = truths[repo]

    for index, output in enumerate(runs, 1):
        checks, _ = score(truth, raw, output)
        rows += [{'model': model, 'repo': repo, 'run': index, **check} for check in checks]

results = pd.DataFrame(rows)
```

```python
rates = results.groupby(['model', 'run'])['passed'].mean()

luna = rates['gpt-6-luna']
haiku = rates['claude-haiku-5-5']

print(f'gpt-6-luna ({len(luna)} runs): mean {luna.mean():.1%}, std {luna.std():.1%} (runs: {", ".join(f"{rate:.1%}" for rate in luna)})')
print(f'claude-haiku-5-5 ({len(haiku)} run): mean {haiku.mean():.1%}')
```
